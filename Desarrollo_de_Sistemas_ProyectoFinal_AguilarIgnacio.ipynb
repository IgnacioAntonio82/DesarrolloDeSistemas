```python


# Cargar el dataset Iris de Scikit-learn.
# as_frame=True permite obtener los datos en formato DataFrame de pandas.
iris = load_iris(as_frame=True)

# Crear una copia del DataFrame original que contiene las características
# de las flores y la variable objetivo codificada numéricamente.
df = iris.frame.copy()

# Crear una nueva columna llamada "species" para mostrar el nombre
# de la especie en lugar de su valor numérico.
# target_names contiene los nombres: setosa, versicolor y virginica.
# enumerate() asocia cada número con el nombre de su especie.
# map() reemplaza los valores de la columna "target" por esos nombres.
df["species"] = df["target"].map(
    dict(enumerate(iris.target_names))
)

# Mostrar las primeras cinco filas del DataFrame para inspeccionar
# las variables, sus valores y la especie correspondiente.
display(df.head())
```

```python
# Descripción de las variables del dataset Iris
variables = {
    "Variable": [
        "sepal length",
        "sepal width",
        "petal length",
        "petal width",
        "species"
    ],
    "Descripción": [
        "Longitud del sépalo de la flor, medida en centímetros.",
        "Ancho del sépalo de la flor, medido en centímetros.",
        "Longitud del pétalo de la flor, medida en centímetros.",
        "Ancho del pétalo de la flor, medido en centímetros.",
        "Especie de la flor. Es la variable objetivo que se busca predecir."
    ],
    "Tipo de variable": [
        "Numérica continua",
        "Numérica continua",
        "Numérica continua",
        "Numérica continua",
        "Categórica"
    ],
    "Rol en el modelo": [
        "Característica de entrada (feature)",
        "Característica de entrada (feature)",
        "Característica de entrada (feature)",
        "Característica de entrada (feature)",
        "Variable objetivo (target)"
    ]
}

df_variables = pd.DataFrame(variables)

display(Markdown("## Descripción de las variables del dataset Iris"))
display(df_variables)
```

```python
# Examinar la estructura y la calidad de los datos (EDA)

# Mostrar la cantidad de filas y columnas del DataFrame.
# Las filas representan flores y las columnas representan variables.
print("Cantidad de filas y columnas:", df.shape)

# Mostrar información general del DataFrame:
# nombres de las columnas, tipos de datos y cantidad de valores no nulos.
# Permite identificar posibles problemas en la estructura de los datos.
print("\nInformación general:")
df.info()

# Calcular estadísticas descriptivas de las cuatro características numéricas:
# media, desviación estándar, mínimo, máximo y cuartiles.
# round(2) redondea los resultados a dos decimales para facilitar la lectura.
# iris.feature_names contiene los nombres de las variables de entrada.
print("\nEstadísticas descriptivas:")
display(df[iris.feature_names].describe().round(2))

# Contar los valores nulos de cada columna.
# isnull() identifica los valores faltantes y sum() cuenta cuántos hay.
# to_frame() presenta el resultado en forma de tabla.
print("\nValores nulos por columna:")
display(df.isnull().sum().to_frame("Valores nulos"))

# Contar las filas duplicadas del DataFrame.
# duplicated() identifica los registros que repiten una fila anterior.
# sum() devuelve la cantidad total de registros duplicados.
print("\nCantidad de filas duplicadas:", df.duplicated().sum())

# Contar cuántas flores hay de cada especie.
# value_counts() calcula la frecuencia de cada categoría.
# to_frame() convierte el resultado en una tabla.
print("\nCantidad de registros por especie:")
display(df["species"].value_counts().to_frame("Cantidad"))
```

```python
plt.figure(figsize=(8, 5))

ax = sns.countplot(
    data=df,
    x="species",
    hue="species",
    order=iris.target_names,
    palette="viridis",
    legend=False
)

ax.set_title("Distribución de las especies de Iris")
ax.set_xlabel("Especie")
ax.set_ylabel("Cantidad de flores")

for container in ax.containers:
    ax.bar_label(container)

plt.tight_layout()
plt.show()
```

```python
#Mostrar las distribuciones de las características
fig, axes = plt.subplots(2, 2, figsize=(12, 8))
axes = axes.flatten()

for i, feature in enumerate(iris.feature_names):
    sns.boxplot(
        data=df,
        x="species",
        y=feature,
        hue="species",
        legend=False,
        ax=axes[i],
        palette="Set2"
    )

    axes[i].set_title(f"Distribución: {feature}")
    axes[i].set_xlabel("Especie")
    axes[i].set_ylabel("Medida (cm)")

plt.tight_layout()
plt.show()
```

El análisis de los diagramas de caja muestra que las características relacionadas con los pétalos permiten diferenciar mejor las especies, mientras que las medidas de los sépalos presentan una mayor superposición entre algunas clases. Por este motivo, utilizar las cuatro variables como entradas de la red neuronal permite aprovechar la información conjunta para mejorar la clasificación de las especies de Iris.

```python
# Preprocesamiento de los datos
# Separar las variables y dividir los datos en entrenamiento y prueba

# Seleccionar las variables predictoras (X).
# Son las cuatro características numéricas que describen cada flor:
# longitud y ancho del sépalo, y longitud y ancho del pétalo.
# copy() crea una copia independiente de los datos seleccionados.
X = df[iris.feature_names].copy()

# Seleccionar la variable objetivo (y).
# Contiene los valores numéricos que identifican cada especie:
# 0 = setosa, 1 = versicolor y 2 = virginica.
# El modelo utilizará estas etiquetas para aprender a clasificar las flores.
y = df["target"].copy()

# Dividir los datos en dos conjuntos:
# - Entrenamiento: 80 % de los registros para entrenar el modelo.
# - Prueba: 20 % de los registros para evaluar su rendimiento.
#
# random_state=42 garantiza que la división sea reproducible.
# stratify=y mantiene aproximadamente la misma proporción de especies
# en los conjuntos de entrenamiento y prueba.
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
# Mostrar la cantidad de registros asignados a cada conjunto.
print("Registros de entrenamiento:", len(X_train))
print("Registros de prueba:", len(X_test))

# Mostrar la cantidad de flores de cada especie en el conjunto
# de entrenamiento. sort_index() ordena las clases por su código numérico.
print("\nDistribución de clases en entrenamiento:")
print(y_train.value_counts().sort_index())

# Mostrar la cantidad de flores de cada especie en el conjunto
# de prueba para verificar que se mantiene la proporción de las clases.
print("\nDistribución de clases en prueba:")
print(y_test.value_counts().sort_index())
```

```python
# Reproducibilidad del experimento
# Configurar las semillas aleatorias para obtener resultados
# consistentes entre distintas ejecuciones.
# Definir una semilla fija para controlar la generación de números aleatorios.
SEED = 42

# Fijar la semilla del generador de números aleatorios de NumPy.
# Esto permite reproducir operaciones aleatorias realizadas con NumPy.
np.random.seed(SEED)
# Fijar las semillas aleatorias utilizadas por TensorFlow y Keras.
# Ayuda a que la inicialización de los pesos y otras operaciones aleatorias
# del modelo produzcan resultados consistentes entre ejecuciones.
tf.keras.utils.set_random_seed(SEED)

# Mostrar la versión de TensorFlow instalada en el entorno de Google Colab.
# Permite identificar la versión de la biblioteca utilizada en el experimento.
print("TensorFlow:", tf.__version__)
```

```python
# Dividir los datos de entrenamiento en entrenamiento y validación

# Separar el conjunto de entrenamiento original en dos subconjuntos:
# - Entrenamiento interno: 80 % de los datos de X_train.
# - Validación: 20 % de los datos de X_train.
#
# random_state=SEED garantiza que la división sea reproducible.
# stratify=y_train mantiene la proporción de las tres especies
# en ambos subconjuntos.
X_subtrain, X_val, y_subtrain, y_val = train_test_split(
    X_train,
    y_train,
    test_size=0.20,
    random_state=SEED,
    stratify=y_train
)

# Estandarizar los datos sin utilizar información de validación o prueba.
# Se crea un estandarizador independiente para esta etapa del experimento.
scaler_search = StandardScaler()

# Calcular la media y la desviación estándar utilizando únicamente
# los datos del subconjunto de entrenamiento y transformarlos.
X_subtrain_scaled = scaler_search.fit_transform(X_subtrain)

# Aplicar a los datos de validación la misma transformación aprendida
# con los datos de entrenamiento, sin recalcular sus parámetros.
X_val_scaled = scaler_search.transform(X_val)

# Mostrar las dimensiones de los conjuntos resultantes.
# Cada fila representa una flor y cada columna una característica.
print("Entrenamiento:", X_subtrain_scaled.shape)
print("Validación:", X_val_scaled.shape)

# Mostrar las dimensiones del conjunto de prueba original,
# que se mantiene reservado para evaluar el modelo al finalizar.
print("Prueba reservada:", X_test.shape)
```

```python
# Función para crear y configurar una red neuronal
# Permite modificar la cantidad de neuronas
#y la tasa de aprendizaje Es un hiperparámetro que determina qué tan grandes son los ajustes que realiza el modelo sobre sus pesos (weights) durante el entrenamiento..
def crear_modelo(neuronas=16, learning_rate=0.001):
   # Crear una red neuronal secuencial, donde las capas se conectan
    # una detrás de otra.
    modelo = keras.Sequential([
        # Definir la capa de entrada.
        # Cada flor se representa mediante 4 características numéricas:
        # largo y ancho del sépalo, y largo y ancho del pétalo.
        keras.Input(shape=(4,)),
        # Primera capa oculta.
        # Contiene una cantidad configurable de neuronas.
        # La función ReLU permite aprender relaciones no lineales
        layers.Dense(neuronas, activation="relu"),
        # Segunda capa oculta.
        # Contiene 8 neuronas y utiliza también la función ReLU
        # para aprender patrones más complejos de los datos.
        layers.Dense(8, activation="relu"),
        # Capa de salida.
        # Tiene 3 neuronas, una por cada especie de Iris. Setosa, Versocolor y Virginica.
        # Softmax genera una probabilidad para cada clase.
        layers.Dense(3, activation="softmax")
    ])
    # Configurar el proceso de aprendizaje de la red neuronal.
    modelo.compile(
        """
          Configuro  tres elementos:
          1. Optimizer: cómo se van a actualizar los pesos de la red neuronal.
          2. Loss: cómo se va a calcular el error de las predicciones.
          3. Metrics: cómo vamos a medir el rendimiento del modelo.
        """
        # Utilizar el optimizador Adam El optimizador es el encargado de modificar los pesos de las neuronas para intentar reducir el error.
        # de la red durante el entrenamiento.
        # learning_rate controla el tamaño de los ajustes realizados.
        optimizer=keras.optimizers.Adam(
            learning_rate=learning_rate
        ),
        # Función de pérdida para clasificación multiclase
        # Esta función calcula qué tan equivocada está la predicción del modelo.
        loss="sparse_categorical_crossentropy",
        # Métrica para medir la proporción de predicciones correctas.
        metrics=["accuracy"]
    )
     # Devolver el modelo creado y compilado para poder entrenarlo.
    return modelo
```

Justificación de la arquitectura:
- Entrada de cuatro valores: las cuatro mediciones de cada flor.
- Primera capa oculta configurable: permite explorar distintas capacidades del modelo.
- Segunda capa de ocho neuronas: aprende combinaciones de las características.
- Salida de tres neuronas con softmax: devuelve probabilidades para las tres especies.
Usamos sparse_categorical_crossentropy porque las etiquetas son números enteros y accuracy porque queremos medir la proporción de clasificaciones correctas.

```python
# Entrenar y evaluar una configuración de la red neuronal

def evaluar_configuracion(neuronas):
    # Limpiar la sesión de Keras para liberar recursos y evitar
    # que se acumulen modelos anteriores en memoria.
    tf.keras.backend.clear_session()
    # Reiniciar las semillas aleatorias para favorecer la reproducibilidad
    # y mejorar la comparabilidad entre distintas configuraciones.
    tf.keras.utils.set_random_seed(SEED)

    # Crear y compilar una red neuronal con la cantidad de neuronas indicada.
    # La tasa de aprendizaje utiliza el valor predeterminado: 0.001.
    modelo = crear_modelo(neuronas=neuronas)
    # Configurar la detención anticipada (Early Stopping).
    # Monitoriza la pérdida de validación (val_loss).
    # patience=12 permite esperar hasta 12 épocas sin mejora antes de detener
    # el entrenamiento.
    # restore_best_weights=True recupera los pesos de la mejor época.
    early_stopping = EarlyStopping(
        monitor="val_loss",
        patience=12,
        restore_best_weights=True
    )

    # Entrenar la red neuronal con los datos de entrenamiento interno.
    historial = modelo.fit(
        X_subtrain_scaled,  # Características de entrenamiento estandarizadas.
        y_subtrain,         # Especies reales de las flores.
        validation_data=(X_val_scaled, y_val),  # Datos para validar el modelo.
        epochs=100,         # Máximo de épocas de entrenamiento.
        batch_size=16,      # Cantidad de registros procesados por lote.
        callbacks=[early_stopping],  # Aplicar detención anticipada.
        verbose=0           # No mostrar el progreso de cada época.
    )

    # Evaluar el modelo con los datos de validación.
    # Devuelve la pérdida y la exactitud utilizando los mejores pesos
    # recuperados por EarlyStopping.
    val_loss, val_accuracy = modelo.evaluate(
        X_val_scaled,
        y_val,
        verbose=0
    )

    # Identificar la época en la que se obtuvo la menor pérdida
    # de validación durante el entrenamiento.
    # argmin() devuelve el índice de la menor pérdida; se suma 1
    # porque las épocas comienzan en 1 para su presentación.
    mejor_epoca = int(np.argmin(historial.history["val_loss"]) + 1)

    # Devolver los resultados de la configuración evaluada.
    # Se incluyen la cantidad de neuronas, las métricas de validación,
    # la mejor época y el historial completo del entrenamiento.
    return {
        "neuronas": neuronas,
        "val_loss": float(val_loss),
        "val_accuracy": float(val_accuracy),
        "mejor_epoca": mejor_epoca,
        "historial": historial
    }
```

```python
# Implementar la búsqueda voraz (Greedy Search)
#
# La búsqueda comienza con 16 neuronas.
# En cada paso, evalúa las configuraciones vecinas y selecciona
# la que produce la menor pérdida de validación.
# Continúa mientras encuentre una mejora estricta.
# Las configuraciones ya evaluadas se guardan para evitar
# repetir entrenamientos innecesarios.


# Definir el espacio de configuraciones posibles.
# Cada valor representa la cantidad de neuronas de la primera capa oculta.
configuraciones = [4, 8, 16, 32, 64]

# Establecer la configuración inicial de la búsqueda.
actual = 16

# Crear un diccionario para almacenar los resultados de cada configuración.
# La cantidad de neuronas será la clave y sus métricas serán los valores.
resultados = {}

# Evaluar una configuración únicamente si todavía no fue evaluada.
def evaluar_si_necesario(neuronas):
    # Comprobar si los resultados de esta configuración ya existen.
    if neuronas not in resultados:
        # Entrenar y evaluar la red neuronal para esta cantidad de neuronas.
        # Guardar las métricas y el historial del entrenamiento.
        resultados[neuronas] = evaluar_configuracion(neuronas)
    # Devolver los resultados almacenados de la configuración.
    return resultados[neuronas]


# Evaluar la configuración inicial de 16 neuronas.
evaluar_si_necesario(actual)

# Repetir la búsqueda hasta que no se encuentre una mejora.
while True:
    # Obtener la posición de la configuración actual dentro de la lista.
    indice = configuraciones.index(actual)

    # Crear una lista para almacenar los vecinos inmediatos.
    vecinos = []
    # Si existe una configuración anterior, agregarla como vecino.
    if indice > 0:
        vecinos.append(configuraciones[indice - 1])

    # Si existe una configuración posterior, agregarla como vecino.
    if indice < len(configuraciones) - 1:
        vecinos.append(configuraciones[indice + 1])

    # Evaluar los vecinos disponibles.
    # Si alguno ya fue evaluado, se reutilizan sus resultados
    for neuronas in vecinos:
        evaluar_si_necesario(neuronas)

    # Construir la lista de opciones que se compararán:
    # la configuración actual y sus vecinos inmediatos.
    opciones = [actual] + vecinos
    # Seleccionar la configuración con la menor pérdida de validación.
    # La función min() compara el valor de val_loss de cada opción.
    mejor_vecino = min(
        opciones,
        key=lambda n: resultados[n]["val_loss"]
    )

    # Detener la búsqueda si la configuración actual sigue siendo
    # la mejor o si ninguna alternativa reduce estrictamente la pérdida.
    if (
        mejor_vecino == actual
        or resultados[mejor_vecino]["val_loss"]
        >= resultados[actual]["val_loss"]
    ):
        print("No se encontró una mejora local.")
        break

    # Mostrar la mejora obtenida al cambiar de configuración.
    print(
        f"Mejora: {actual} -> {mejor_vecino} neuronas | "
        f"Pérdida de validación: "
        f"{resultados[actual]['val_loss']:.4f} -> "
        f"{resultados[mejor_vecino]['val_loss']:.4f}"
    )
    # Actualizar la configuración actual para continuar la búsqueda
    # desde la alternativa que obtuvo la menor pérdida.
    actual = mejor_vecino


# Construir una tabla con todas las configuraciones evaluadas.
# Incluir la cantidad de neuronas, la pérdida, la exactitud
# y la época con menor pérdida de validación.
tabla_resultados = pd.DataFrame([
    {
        "Neuronas": r["neuronas"],
        "Pérdida de validación": r["val_loss"],
        "Accuracy de validación": r["val_accuracy"],
        "Mejor época": r["mejor_epoca"]
    }
    for r in resultados.values()
]).sort_values("Neuronas")

# Mostrar los resultados redondeados a cuatro decimales.
display(tabla_resultados.round(4))
# Recuperar los resultados de la configuración seleccionada.
mejor_configuracion = resultados[actual]
# Mostrar la configuración final y su pérdida de validación.
print("Configuración seleccionada:", actual, "neuronas")
print(
    "Pérdida de validación:",
    round(mejor_configuracion["val_loss"], 4)
)
```

### ¿Por qué esto es una búsqueda voraz?

En cada iteración, el algoritmo selecciona la configuración que presenta la menor pérdida de validación entre la configuración actual y sus vecinos inmediatos.

- **Decisión local:** elige la alternativa que ofrece el mejor resultado en el entorno inmediato.
- **Criterio de selección:** utiliza la pérdida de validación (`val_loss`) para comparar las configuraciones.
- **Condición de parada:** finaliza cuando no encuentra una configuración vecina que reduzca estrictamente la pérdida.
- **Limitación:** no evalúa necesariamente todas las configuraciones posibles, por lo que puede encontrar un óptimo local sin garantizar el óptimo global.

En este proyecto, la búsqueda voraz permite seleccionar la cantidad de neuronas de la primera capa oculta de la red neuronal. Esta estrategia ilustra cómo una técnica clásica de búsqueda puede aplicarse a la selección de hiperparámetros y se diferencia de una búsqueda exhaustiva, que evalúa todas las configuraciones del espacio definido.

## ¿Cómo funciona la búsqueda voraz?
-  Comenzar en 16 neuronas

    - Se entrena el modelo inicial y se guarda su pérdida de validación.


- Evaluar los vecinos

    - Se prueban las configuraciones inmediatamente anteriores y posteriores, cuando existen.


- Elegir la mejor alternativa

    - Se avanza a la configuración con menor pérdida, siempre que mejore el resultado actual.


-  Detener la búsqueda

    - El proceso termina cuando ningún vecino ofrece una mejora estricta.

```python
# Entrenamiento final y evaluación en el conjunto de prueba (test)

# Crear un nuevo estandarizador para el modelo final.
scaler_final = StandardScaler()

# Ajustar el escalador con todo el conjunto de entrenamiento.
# fit_transform() calcula la media y la desviación estándar
# de las 4 características y transforma los 120 registros.
X_train_final = scaler_final.fit_transform(X_train)
# Aplicar a los datos de prueba la misma transformación aprendida
# con el conjunto de entrenamiento.
# No se calculan nuevas estadísticas con los datos de prueba.
X_test_final = scaler_final.transform(X_test)

# Crear la red neuronal con la cantidad de neuronas seleccionada
# durante la búsqueda voraz.
# Limpiar la sesión de Keras antes de crear el nuevo modelo.
tf.keras.backend.clear_session()
# Reiniciar las semillas aleatorias para favorecer la reproducibilidad.
tf.keras.utils.set_random_seed(SEED)

# Construir y compilar el modelo final.
modelo_final = crear_modelo(neuronas=actual)

# Utilizar como referencia la mejor época observada durante
# la validación de la configuración seleccionada.
# max(1, ...) garantiza que se entrene al menos una época.
epocas_finales = max(
    1, mejor_configuracion["mejor_epoca"]
)

# Entrenar el modelo con todo el conjunto de entrenamiento.
# X_train_final contiene las características estandarizadas.
# y_train contiene las etiquetas reales de las especies (0, 1 y 2).
# epochs establece la cantidad de épocas de entrenamiento.
# batch_size=16 indica que se procesan 16 registros por lote.
# verbose=0 oculta el progreso de cada época.
historial_final = modelo_final.fit(
    X_train_final,
    y_train,
    epochs=epocas_finales,
    batch_size=16,
    verbose=0
)

# Evaluar el modelo final utilizando los datos de prueba reservados.
# test_loss mide el error de las predicciones.
# test_accuracy mide la proporción de clasificaciones correctas.
test_loss, test_accuracy = modelo_final.evaluate(
    X_test_final,
    y_test,
    verbose=0
)

# Obtener las probabilidades predichas para cada una de las
# tres especies de Iris en cada registro del conjunto de prueba.
# El resultado tiene una columna por especie.
y_probabilidades = modelo_final.predict(
    X_test_final,
    verbose=0
)

# Convertir las probabilidades en clases predichas.
# argmax(axis=1) selecciona, para cada flor, la especie
# cuya probabilidad predicha es la más alta.
# Las clases resultantes son 0, 1 o 2.
y_pred = np.argmax(y_probabilidades, axis=1)
```

```python
# Mostrar las métricas finales del modelo sobre el conjunto de prueba (test).

# Imprimir un encabezado para identificar la sección de evaluación.
print("===== EVALUACIÓN FINAL EN TEST =====")
# Mostrar la pérdida (loss), que mide el error de las predicciones.
# Cuanto menor sea su valor, mejor será el resultado según esta métrica.
print(f"Pérdida (loss): {test_loss:.4f}")
# Mostrar la exactitud (accuracy), que representa la proporción
# de flores clasificadas correctamente.
# El valor se presenta con cuatro decimales.
print(f"Accuracy: {test_accuracy:.4f}")

# Generar un informe detallado del rendimiento del modelo por especie.
print("\n===== INFORME DE CLASIFICACIÓN =====")
# Comparar las especies reales (y_test) con las especies predichas (y_pred).
# labels establece el orden de las clases: 0, 1 y 2.
# target_names muestra los nombres de las especies en lugar de sus códigos.
# zero_division=0 evita advertencias y devuelve 0 cuando una métrica
# no puede calcularse por ausencia de predicciones o casos relevantes.
print(classification_report(
    y_test,
    y_pred,
    labels=[0, 1, 2],
    target_names=iris.target_names,
    zero_division=0
))

# Calcular la matriz de confusión para comparar las clases reales
# con las clases predichas por el modelo.
matriz = confusion_matrix(
    y_test,
    y_pred,
    labels=[0, 1, 2]
)
# Crear un objeto para visualizar la matriz de confusión.
# Los nombres de las especies se utilizan como etiquetas de los ejes.
disp = ConfusionMatrixDisplay(
    confusion_matrix=matriz,
    display_labels=iris.target_names
)

# Dibujar la matriz con una escala de colores azules.
# values_format="d" muestra las cantidades como números enteros.
disp.plot(cmap="Blues", values_format="d")
# Agregar un título al gráfico.
plt.title("Matriz de confusión - Test")
# Ajustar los márgenes para mejorar la presentación.
plt.tight_layout()
# Mostrar la matriz de confusión.
plt.show()
```

# Razonamiento lógico basado en reglas

```python
# ==========================================
# MODULO 3: RAZONAMIENTO LOGICO Y REGLAS
# ==========================================

# Este modulo utiliza reglas explicitas para evaluar
# si las caracteristicas de una flor son compatibles
# con una determinada especie.

def motor_reglas(flor):

    # Obtener las medidas originales en centimetros.
    longitud_petalo = flor[0][2]
    ancho_petalo = flor[0][3]

    # Regla 1: caracteristicas tipicas de Setosa.
    if longitud_petalo <= 2.0 and ancho_petalo <= 0.6:
        return "setosa", "R1: Petalos pequenos"

    # Regla 2: caracteristicas orientativas de Versicolor.
    elif (longitud_petalo > 2.0 and
          longitud_petalo <= 4.9 and
          ancho_petalo <= 1.7):
        return "versicolor", "R2: Petalos intermedios"

    # Regla 3: caracteristicas orientativas de Virginica.
    elif longitud_petalo > 4.9 and ancho_petalo > 1.4:
        return "virginica", "R3: Petalos grandes"

    # Si ninguna regla se cumple, no se fuerza una conclusion.
    else:
        return None, "R4: No se encontro una regla concluyente"


# Ejecutar el motor de reglas sobre la flor ingresada.
especie_regla, regla_aplicada = motor_reglas(nueva_flor)

# Obtener la especie predicha por la red neuronal.
especie_ml = iris.target_names[clase]

# Mostrar los resultados de ambos componentes.
print("===== RAZONAMIENTO LOGICO =====")
print("Regla aplicada:", regla_aplicada)

print("\n===== COMPARACION DE RESULTADOS =====")
print("Prediccion Machine Learning:", especie_ml)

if especie_regla is not None:
    print("Inferencia del motor de reglas:", especie_regla)

    if especie_ml == especie_regla:
        print("Resultado: AMBOS COMPONENTES COINCIDEN")
    else:
        print("Resultado: DISCREPANCIA ENTRE ML Y REGLAS")

else:
    print("Inferencia del motor de reglas: INCONCLUSA")
    print("La red neuronal mantiene su prediccion.")

print("\n===== RESULTADO FINAL =====")
print("Especie predicha:", especie_ml)
```
